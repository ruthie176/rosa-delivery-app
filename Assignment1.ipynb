## Part 1

#### Part 1a

```python
def percentage_lateorders(z_list, t_list, promise):

    if z_list == 'all':
        zones = ZONES
    else:
        zones = [z_list]

    if t_list == 'all':
        blocks = TIME_BLOCKS
    
    else:
        blocks = [t_list]

    times = []

    for z in zones:
        for b in blocks:
            new_times = delivery_times(z, b, promise,seed=1)
            for t in new_times:
                times.append(t)

    late = 0
    for t in times:
        if t > promise:
            late += 1

    total = len(times)
    percentage = late / total * 100

    return percentage
```

#### Part 1b

```python
results = [] # Creating a list for the results so I can rank them

for Z in ZONES:
    for T in TIME_BLOCKS:
        rate = percentage_lateorders(Z,T,PROMISE)
        results.append((Z,T,rate)) # This adds each loop to results

ranked = sorted(results, key=lambda x: x[2], reverse=True) # This code is to sort the results based on the rate from high to low

for rank, (zone, block, rate) in enumerate(ranked, start=1):
    print(f"{rank}. {zone} - {block}: {rate:.0f}%")
```

#### Part 1c

```python
def avg_deliverytime(z_list, t_list, promise):

    if z_list == 'all':
        zones = ZONES
    else:
        zones = [z_list]

    if t_list == 'all':
        blocks = TIME_BLOCKS
    
    else:
        blocks = [t_list]

    times = []

    for z in zones:
        for b in blocks:
            new_times = delivery_times(z, b, promise, seed=1)
            for t in new_times:
                times.append(t)

    return round(float(np.mean(times)))
```

##### The code to calculate the total cost per late order is shown below

```python
costs = COSTS
def total_cost(costs):
    totalcost = costs['refund'] + (costs['churn_orders'] * costs['margin'])

    return(totalcost)
```

#### Part 2b

```python
def net_profit(z, t_b, p, costs):
    order_times = delivery_times(z, t_b, p, seed=1)
    late_orders = np.sum(order_times > p)           
    sales = costs['margin'] * len(order_times)     
    late_costs = late_orders * total_cost(costs)
    return sales - late_costs
```

##### The codes below shows the statistics for a zone: Far West and time block: Fri/Sat eve. This will help to see where most of the deliveries fall and to determine a best promise time

```python
t = delivery_times('Far West', 'Fri/Sat eve', PROMISE, seed=1) 

print(f"Minimum:          {np.min(t):.1f} min")
print(f"10th percentile:  {np.percentile(t, 10):.1f} min")
print(f"Average:          {np.mean(t):.1f} min")
print(f"90th percentile:  {np.percentile(t, 90):.1f} min")
print(f"Maximum:          {np.max(t):.1f} min")
```

```python
promises = list(range(5, 55, 5)) # This increases the promise time in steps of 5 and stops at 55
table = []

print(f"{'Promise':>8} {'Orders':>8} {'Late %':>8} {'Net profit':>12}")
for p in promises:
    t = delivery_times('Far West', 'Fri/Sat eve', p, seed=1)
    profit = net_profit('Far West', 'Fri/Sat eve', p, COSTS)
    table.append((p, profit))
    print(f"{p:>8} {len(t):>8} {np.mean(t > p) * 100:>7.0f}% {profit:>12,.2f}")
```

##### Using 'Far West' and 'Fri/Sat eve', the statistics above shows that the fastest delivery takes 18.6 minutes and only 10% of orders arrive within 29.3 minutes or less. Promise time shorter than about 25 minutes are therefore unrealistic because over 97% of orders would be late, and Rosa would lose more than $3,900. This is why the lower end of the range does not need to go below 25 minutes, and the useful promises start closer to 40 minutes.

```python
promises = list(range(5, 85, 5))

def best_promise(z, t_b, promise_list, costs):
       best_p = None
       best_profit = float('-inf')
       for p in promise_list:
           profit = net_profit(z, t_b, p, costs)
           if profit > best_profit:
               best_profit = profit
               best_p = p
       return best_p, best_profit
```

##### The range chosen is from 5 to 80 minutes because. This is because the lower end is kept low so the function also works for faster zones and time blocks. The upper end of the range (from 60 - 80) is well beyond the 90th percentile of delivery times for the slowest pair (55.4 minutes for Far West, Fri/Sat eve). The edge check confirms that no pair's best promise falls at either end of the range.

```python
p, profit = best_promise('Far West','Fri/Sat eve', promises, COSTS)
print(f"Recommended promise: {p} minutes")
print(f"Net profit: ${profit:,.2f}")
```

##### For zone Far West, during Fri/Sat eve time block, the best promise time was calculated to be 55 minutes. 

```python
for z in ZONES:
    for b in TIME_BLOCKS:
        p, profit = best_promise(z, b, promises, COSTS)
        if p == promises[0] or p == promises[-1]:
            note = "AT EDGE: widen the range"
        else:
            note = "inside range: OK"
        print(f"{z} - {b}: best {p} min, net profit ${profit:,.2f} ({note})")
```

##### All 12 zone and time block pairs have their best promise inside the 5 – 80 minute range, so the range is wide enough to capture the most profitable promise for every pair.
